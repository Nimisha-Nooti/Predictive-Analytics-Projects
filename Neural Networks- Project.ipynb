### Illustrating a Single Forward Pass in a Neural Network with Randomly Initialized Weights and Biases

### Import necessary libraries

```python
import numpy as np
import pandas as pd
```

### Convert the data to a data frame

```python
# Define the data 
data = {
    "Years": [4, 18, 1, 3, 15, 6],
    "Salary": [43, 65, 53, 95, 88, 112],
    "UsedCredit": [0, 1, 0, 0, 1, 1]
}

df = pd.DataFrame(data)

# Display the data
print("Given Data:")
print(df)
```

### Check for missing values

```python
missing_values = df.isnull().sum()
print("\nMissing values in the dataset:")
print(missing_values)
```

### Without Normalisation

```python
# Defined activation function (sigmoid)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Randomly generate initial weights and biases
np.random.seed(42)
weights = np.random.rand(2, 1)  # 2 inputs, 1 output
bias = np.random.rand(1)

# Extract inputs from the DataFrame
inputs = df[["Years", "Salary"]].values

# Forward pass: Compute weighted sum and apply activation
weighted_sum = np.dot(inputs, weights) + bias
predictions = sigmoid(weighted_sum)

df["Predictions"] = predictions

print("Randomly initialized weights:")
print(weights)

print("\nRandomly initialized bias:")
print(bias)

print("\nData with Predictions:")
print(df)
```

### With Normalisation

```python
# Define activation function (sigmoid)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Randomly generate initial weights and biases
np.random.seed(42)
weights = np.random.rand(2, 1)  # 2 inputs, 1 output
bias = np.random.rand(1)


# Extract inputs from the DataFrame
inputs = df[["Years", "Salary"]].values

# Normalize inputs to bring values to a smaller range
inputs = inputs / np.max(inputs, axis=0)

# Forward pass: Compute weighted sum and apply activation
weighted_sum = np.dot(inputs, weights) + bias
predictions = sigmoid(weighted_sum)

df["Predictions"] = predictions

# Display the results
print("Randomly initialized weights:")
print(weights)

print("\nRandomly initialized bias:")
print(bias)

print("\nData with Predictions (Normalized Inputs):")
print(df)


```

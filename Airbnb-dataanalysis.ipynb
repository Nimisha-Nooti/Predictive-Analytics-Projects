## Check for outliers

### Function to Identify Outliers Using the IQR Method

```python
def custom_finder(airbnbdf):
   q1=airbnbdf.quantile(0.25)
   q3=airbnbdf.quantile(0.75)
   IQR=q3-q1
   out_l = airbnbdf[((airbnbdf<(q1-1.5*IQR)) | (airbnbdf>(q3+1.5*IQR)))]
   return out_l
```

```python
outl = custom_finder(airbnbdf['price'])
print("Number of outliers in the 'price' column: "+ str(len(outl)))
```

```python
outl = custom_finder(airbnbdf['beds'])
print("Number of outliers in the 'beds' column: "+ str(len(outl)))
```

```python
outl = custom_finder(airbnbdf['bedrooms'])
print("Number of outliers in the 'bedrooms' column: "+ str(len(outl)))
```

```python
outl = custom_finder(airbnbdf['review_scores_rating'])
print("Number of outliers in the 'review_scores_rating' column: "+ str(len(outl)))
```

### Handling missing values in the 'review_scores_rating' column by replacing them with median values

Since there are many rows with missing values in the 'review_scores_rating' column, dropping them could affect the integrity of the dataset. Hence, replacing the missing values with the median is a more appropriate approach to handle the inconsistency.

```python
# Calculate the median of 'review_scores_rating'
median_review_score = airbnbdf['review_scores_rating'].median()
```

```python
median_review_score
```

```python
# Replace NA values with the median
airbnbdf['review_scores_rating'].fillna(median_review_score, inplace=True)
```

```python
# Verify if all missing values were replaced
remaining_na_values = airbnbdf['review_scores_rating'].isna().sum()
print(f"Remaining missing values after replacement: {remaining_na_values}")
```

```python
# Print rows where the value was replaced by the median
replaced_rows = airbnbdf[airbnbdf['review_scores_rating'] == median_review_score]
print("Rows where missing values were replaced with the median:")
print(replaced_rows[['review_scores_rating']])
```
